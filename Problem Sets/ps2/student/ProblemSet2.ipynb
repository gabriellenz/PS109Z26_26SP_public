## Problem Set 2: Death Sentence and Discrimination

In a 1987 Supreme Court case, *McCleskey v. Kemp*, McCleskey's lawyers argued that the race of both the victim and the defendant played a major role in whether someone received the death penalty, even after accounting for other facts of the case. They argued that the death penalty was therefore being applied in an unconstitutionally discriminatory way.

In a 5-4 decision, the Supreme Court rejected McCleskey's claim. Justice Powell wrote for the majority that broad statistical evidence was not enough by itself. Instead, McCleskey would have to show strong evidence of racial bias in his own case. Powell was later quoted as saying that he eventually changed his mind and would have voted differently.

The legal issue did not disappear. In 1994, members of the House of Representatives proposed allowing statistical evidence to be used to challenge death sentences, but that provision was removed from the final bill after disagreement between the House and Senate.

In this problem set, you will reanalyze data from the McCleskey case. The file `death.rds` contains 98 cases used as part of McCleskey's evidence.

Please run the code in the cell below.
